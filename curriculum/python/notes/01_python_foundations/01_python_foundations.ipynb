## Section 1: Python Syntax Essentials & Dynamic Typing

### 1.1 Indentation IS the Syntax
In Java/C# you use `{}` to define blocks. In Python, **indentation** (4 spaces) defines blocks.

```python
# ✅ Correct — 4 spaces per indentation level
def train_one_epoch(model, dataloader, optimizer, criterion):
    """Train the model for one epoch and return average loss."""
    print("Training started...")
    total_loss = 0.0
    for batch_idx in range(5):  # simulating batches
        loss = 1.0 / (batch_idx + 1)  # simulated decreasing loss
        total_loss += loss
        print(f"  Batch {batch_idx}: loss = {loss:.4f}")
    avg_loss = total_loss / 5
    print(f"Average loss: {avg_loss:.4f}")
    return avg_loss

# Call the function
train_one_epoch(None, None, None, None)
```

### 1.2 Key Differences from Java/C#

|---------|---------|--------|
| Block delimiters | `{ }` | Indentation (4 spaces) |
| Line terminator | `;` | Newline |
| Comments | `//` and `/* */` | `#` only |
| Variable declaration | `int x = 5;` | `x = 5` |
| Constants | `final int X = 5;` | `X = 5` (convention: UPPER_CASE) |
| Null | `null` | `None` |
| Boolean | `true` / `false` | `True` / `False` (capitalized!) |

### 1.5 Constants in Python

**Python has no built-in `const` or `final` keyword.** Unlike Java's `final` or C#'s `const`, there's no way to make a variable truly immutable at the language level. Here are three approaches:

```python
# === Approach 1: UPPER_CASE Convention (99% of Python code) ===
# Just name it in UPPER_CASE — everyone agrees not to reassign
MAX_SEQ_LENGTH = 512
LEARNING_RATE = 3e-4
PAD_TOKEN_ID = 0
D_MODEL = 768
NUM_HEADS = 12

print(f"MAX_SEQ_LENGTH = {MAX_SEQ_LENGTH}")
print(f"D_MODEL = {D_MODEL}")

# ⚠️ Nothing stops you from reassigning!
MAX_SEQ_LENGTH = 1024  # Works! Python won't complain.
print(f"\nAfter reassignment: MAX_SEQ_LENGTH = {MAX_SEQ_LENGTH}")
print("Python trusts the programmer — UPPER_CASE is just a convention.")
```

```python
# === Approach 2: Final type hint (Python 3.8+) — IDE warning only ===
from typing import Final

MAX_SEQ_LENGTH: Final = 512
LEARNING_RATE: Final[float] = 3e-4
MODEL_NAME: Final[str] = "bert-base-uncased"

print(f"MAX_SEQ_LENGTH: Final = {MAX_SEQ_LENGTH}")
print(f"MODEL_NAME: Final = {MODEL_NAME}")

# IDE shows a warning ⚠️, mypy flags it, but Python allows it at runtime
MAX_SEQ_LENGTH = 1024  # Your IDE will underline this, but it runs fine
print(f"After reassign: {MAX_SEQ_LENGTH} — no runtime error!")
```

```python
# === Approach 3: True enforcement via __setattr__ ===
class _Constants:
    """Raises an error if you try to reassign an attribute."""
    def __setattr__(self, name, value):
        if name in self.__dict__:
            raise AttributeError(f"Cannot reassign constant '{name}'")
        super().__setattr__(name, value)

CONST = _Constants()
CONST.MAX_SEQ_LENGTH = 512
CONST.PAD_TOKEN_ID = 0

print(f"CONST.MAX_SEQ_LENGTH = {CONST.MAX_SEQ_LENGTH}")
print(f"CONST.PAD_TOKEN_ID = {CONST.PAD_TOKEN_ID}")

# Uncomment the next line to see the error:
# CONST.MAX_SEQ_LENGTH = 1024  # ❌ AttributeError!

try:
    CONST.MAX_SEQ_LENGTH = 1024
except AttributeError as e:
    print(f"\n❌ Caught: {e}")
    print("This is the ONLY way to truly prevent reassignment in Python!")
```

### ⚠️ Section 1 Pitfalls

```python
# PITFALL 1: Mutable default arguments
# ❌ NEVER use a mutable default
def add_layer_BROKEN(layers=[]):
    layers.append("linear")
    return layers

print("BROKEN version:")
print(f"  Call 1: {add_layer_BROKEN()}")  # ['linear']
print(f"  Call 2: {add_layer_BROKEN()}")  # ['linear', 'linear'] ← BUG!

# ✅ Use None instead
def add_layer_FIXED(layers=None):
    if layers is None:
        layers = []
    layers.append("linear")
    return layers

print("\nFIXED version:")
print(f"  Call 1: {add_layer_FIXED()}")  # ['linear']
print(f"  Call 2: {add_layer_FIXED()}")  # ['linear'] ← Correct!
```

```python
# PITFALL 2: `is` vs `==`
a = [1, 2, 3]
b = [1, 2, 3]
print(f"a == b: {a == b}")   # True — same VALUE
print(f"a is b: {a is b}")   # False — different OBJECTS

# Only use `is` for None, True, False

# PITFALL 3: ^ is XOR, not power
print(f"\n2 ^ 3 = {2 ^ 3}")    # 1 (bitwise XOR)
print(f"2 ** 3 = {2 ** 3}")    # 8 (exponentiation)
```

## Section 3: Type Hints & Annotations

Type hints make code **readable and self-documenting**. They don't enforce types at runtime.

```python
# Variable annotations
model_name: str = "bert-base"
num_layers: int = 12
learning_rate: float = 3e-4
is_training: bool = True

# Function annotations
def compute_loss(predictions: list, targets: list) -> float:
    """Compute mean squared error loss."""
    total = sum((p - t) ** 2 for p, t in zip(predictions, targets))
    return total / len(predictions)

result: float = compute_loss([1.0, 2.0, 3.0], [1.1, 1.9, 3.2])
print(f"MSE Loss: {result:.6f}")
```

```python
# Collection type hints (Python 3.9+)
scores: list[float] = [0.95, 0.87, 0.92]
config: dict[str, int] = {"d_model": 512, "nhead": 8}
unique_tokens: set[str] = {"the", "a", "an"}
shape: tuple[int, int, int] = (32, 128, 512)

print(f"Scores: {scores}")
print(f"Config: {config}")
print(f"Tokens: {unique_tokens}")
print(f"Shape:  {shape}")
```

```python
from typing import Optional, Union, Any, Callable

# Optional[X] = X or None — THE most common type hint in ML code
def create_model(
    d_model: int = 512,
    dropout: float = 0.1,
    pretrained_path: Optional[str] = None,
    custom_head: Optional[Callable] = None
) -> str:
    result = f"Model(d_model={d_model}, dropout={dropout})"
    if pretrained_path:
        result += f" + pretrained({pretrained_path})"
    if custom_head:
        result += f" + custom_head"
    return result

print(create_model())
print(create_model(pretrained_path="model.pt"))

# Union[X, Y] = X or Y
def set_device(device: Union[str, int]) -> str:
    if isinstance(device, int):
        return f"cuda:{device}"
    return device

print(f"\nDevice: {set_device('cpu')}")
print(f"Device: {set_device(0)}")
```

### 5.4 `while` Loop — Early Stopping Pattern

```python
import random
random.seed(42)

# Early stopping pattern — very common in ML training
patience = 3
best_loss = float('inf')
epochs_without_improvement = 0
epoch = 0

while epochs_without_improvement < patience and epoch < 20:
    epoch += 1
    # Simulate a loss that mostly decreases but sometimes doesn't
    current_loss = 3.0 / epoch + random.gauss(0, 0.3)
    current_loss = max(0.01, current_loss)

    if current_loss < best_loss:
        best_loss = current_loss
        epochs_without_improvement = 0
        print(f"Epoch {epoch:2d}: loss={current_loss:.4f} ✅ New best!")
    else:
        epochs_without_improvement += 1
        print(f"Epoch {epoch:2d}: loss={current_loss:.4f} ⏳ No improvement "
              f"({epochs_without_improvement}/{patience})")

print(f"\nEarly stopping at epoch {epoch}. Best loss: {best_loss:.4f}")
```

## Section 6: Comprehensions

Python's most powerful one-liner tool — used on **every page** of AI code.

### 6.1 List Comprehensions

```python
# Basic: [expression FOR variable IN iterable]
squares = [x ** 2 for x in range(10)]
print(f"Squares: {squares}")

# With filter: [expression FOR var IN iterable IF condition]
even_squares = [x ** 2 for x in range(10) if x % 2 == 0]
print(f"Even squares: {even_squares}")

# Transform + filter
words = ["the", "transformer", "is", "great"]
long_word_lengths = [len(w) for w in words if len(w) > 2]
print(f"Long word lengths: {long_word_lengths}")
```

```python
# 🔥 AI Context: Comprehensions in real ML code

# Building vocabulary from tokens
tokens = ["the", "cat", "sat", "on", "the", "mat"]
vocab = sorted(set(tokens))
token_to_idx = {token: idx for idx, token in enumerate(vocab)}
print(f"Vocab: {token_to_idx}")

# Encoding a sentence
sentence = "the cat sat"
encoded = [token_to_idx[word] for word in sentence.split()]
print(f"Encoded: {encoded}")

# Filtering layers (common in fine-tuning)
all_params = ["encoder.layer.0.weight", "encoder.layer.0.bias",
              "classifier.weight", "classifier.bias"]
classifier_params = [p for p in all_params if "classifier" in p]
print(f"Classifier params: {classifier_params}")

# Flatten list of lists
batches = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flat = [item for batch in batches for item in batch]
print(f"Flattened: {flat}")
```

### 6.2 Dict Comprehensions

```python
# {key_expr: value_expr FOR var IN iterable}
words = ["attention", "is", "all", "you", "need"]
word_lengths = {word: len(word) for word in words}
print(f"Word lengths: {word_lengths}")

# Inverting a dictionary
token_to_id = {"[PAD]": 0, "[UNK]": 1, "[CLS]": 2, "[SEP]": 3}
id_to_token = {idx: token for token, idx in token_to_id.items()}
print(f"Inverted: {id_to_token}")

# Filtering a dict
config = {"d_model": 512, "nhead": 8, "dropout": 0.1, "num_layers": 6}
large_params = {k: v for k, v in config.items() 
                if isinstance(v, (int, float)) and v > 10}
print(f"Large params: {large_params}")
```

```python
# Basic unpacking
x, y, z = 1, 2, 3
print(f"x={x}, y={y}, z={z}")

# Swap variables — no temp needed!
a, b = 1, 2
a, b = b, a
print(f"After swap: a={a}, b={b}")

# Tuple unpacking
point = (3.0, 4.0)
x, y = point
print(f"Point: x={x}, y={y}")

# Works with any iterable
a, b, c = "xyz"
print(f"a={a}, b={b}, c={c}")
```

```python
# Star unpacking (*) — captures "the rest"
first, *rest = [1, 2, 3, 4, 5]
print(f"first={first}, rest={rest}")

first, *middle, last = [1, 2, 3, 4, 5]
print(f"first={first}, middle={middle}, last={last}")

# Ignoring values with _
_, important, _ = (1, 42, 3)
print(f"important={important}")

first, *_ = [1, 2, 3, 4, 5]
print(f"first={first} (rest discarded)")
```

## Section 8: Walrus Operator (`:=`)

Assignment expressions (Python 3.8+) — assign a value **as part of an expression**.

```python
# Without walrus — compute len() twice
data = ["transformer", "is", "great", "at", "nlp"]
long_lengths_v1 = []
for word in data:
    if len(word) > 3:
        long_lengths_v1.append(len(word))  # Recomputing len()!
print(f"Without walrus: {long_lengths_v1}")

# With walrus — compute once, use twice
long_lengths_v2 = [length for word in data if (length := len(word)) > 3]
print(f"With walrus:    {long_lengths_v2}")
```

## 🧪 Practice Challenge: Training Dashboard Simulator

This exercise combines **ALL** concepts from Module 1!

```python
"""
Practice Challenge: Build a Training Dashboard Simulator
=========================================================
Uses: variables, types, f-strings, type hints, conditionals,
      loops, comprehensions, unpacking, ternary, walrus, debugging
"""

from typing import Optional
import random
import math

# ===== 1. Configuration (type hints, variables, ternary) =====
class TrainingConfig:
    def __init__(
        self,
        model_name: str = "MiniTransformer",
        num_epochs: int = 10,
        batch_size: int = 32,
        learning_rate: float = 3e-4,
        use_gpu: bool = True,
        weight_decay: float = 0.01,
        patience: int = 3,
        warmup_steps: Optional[int] = None,
    ):
        self.model_name = model_name
        self.num_epochs = num_epochs
        self.batch_size = batch_size
        self.learning_rate = learning_rate
        self.device = "cuda" if use_gpu else "cpu"  # ternary
        self.weight_decay = weight_decay
        self.patience = patience
        self.warmup_steps = warmup_steps if warmup_steps is not None else num_epochs // 5

    def __repr__(self) -> str:
        return (f"TrainingConfig(model={self.model_name!r}, "
                f"epochs={self.num_epochs}, bs={self.batch_size}, "
                f"lr={self.learning_rate:.1e}, device={self.device!r})")

    def __str__(self) -> str:
        return f"{self.model_name} on {self.device} for {self.num_epochs} epochs"


# ===== 2. Simulated Training (loops, f-strings, conditionals) =====
def simulate_training(config: TrainingConfig) -> dict[str, list[float]]:
    """Simulate a training run and return metrics history."""

    print(f"\n{'='*60}")
    print(f"  🚀 Starting Training: {config}")
    print(f"{'='*60}")
    print(f"  Config: {config!r}")
    print()

    # Log all hyperparameters (dict iteration)
    hyperparams = {
        "Model": config.model_name,
        "Device": config.device,
        "Epochs": config.num_epochs,
        "Batch Size": config.batch_size,
        "Learning Rate": config.learning_rate,
        "Weight Decay": config.weight_decay,
        "Warmup Steps": config.warmup_steps,
        "Patience": config.patience,
    }

    print("  Hyperparameters:")
    for key, value in hyperparams.items():
        formatted = f"{value:.1e}" if isinstance(value, float) and value < 0.01 else str(value)
        print(f"    {key:>15}: {formatted}")
    print()

    # Training loop
    history: dict[str, list[float]] = {"loss": [], "accuracy": []}
    best_loss = float('inf')
    epochs_without_improvement = 0

    random.seed(42)

    for epoch in range(1, config.num_epochs + 1):
        # Simulate metrics
        base_loss = 3.0 * math.exp(-0.3 * epoch) + 0.1
        noise = random.gauss(0, 0.05)
        epoch_loss = max(0.01, base_loss + noise)

        base_acc = 1.0 - math.exp(-0.4 * epoch)
        acc_noise = random.gauss(0, 0.02)
        epoch_acc = min(0.99, max(0.0, base_acc + acc_noise))

        history["loss"].append(epoch_loss)
        history["accuracy"].append(epoch_acc)

        # Learning rate warmup (walrus operator)
        if (effective_lr := config.learning_rate * min(1.0, epoch / max(1, config.warmup_steps))) < config.learning_rate:
            lr_status = f"(warmup: {effective_lr:.1e})"
        else:
            lr_status = ""

        # Status emoji (chained conditionals)
        if epoch_loss < 0.1:
            status = "🎯"
        elif epoch_loss < 0.5:
            status = "✅"
        elif epoch_loss < 1.0:
            status = "📈"
        else:
            status = "⏳"

        print(f"  Epoch [{epoch:3d}/{config.num_epochs}] {status} "
              f"Loss: {epoch_loss:.4f} | "
              f"Acc: {epoch_acc:.2%} | "
              f"LR: {effective_lr:.1e} {lr_status}")

        # Early stopping check
        if epoch_loss < best_loss:
            best_loss = epoch_loss
            epochs_without_improvement = 0
        else:
            epochs_without_improvement += 1
            if epochs_without_improvement >= config.patience:
                print(f"\n  ⚠️  Early stopping at epoch {epoch} "
                      f"(no improvement for {config.patience} epochs)")
                break

    return history


# ===== 3. Analysis (comprehensions, unpacking, zip) =====
def analyze_results(history: dict[str, list[float]], config: TrainingConfig) -> None:
    """Analyze and display training results."""

    losses = history["loss"]
    accuracies = history["accuracy"]

    print(f"\n{'='*60}")
    print(f"  📊 Training Analysis")
    print(f"{'='*60}\n")

    best_loss = min(losses)
    best_acc = max(accuracies)
    best_loss_epoch = losses.index(best_loss) + 1
    best_acc_epoch = accuracies.index(best_acc) + 1

    print(f"  Best Loss:     {best_loss:.4f} (epoch {best_loss_epoch})")
    print(f"  Best Accuracy: {best_acc:.2%} (epoch {best_acc_epoch})")
    print(f"  Final Loss:    {losses[-1]:.4f}")
    print(f"  Final Acc:     {accuracies[-1]:.2%}")

    # Epoch-over-epoch improvement (zip)
    improvements = [prev - curr for prev, curr in zip(losses[:-1], losses[1:])]
    positive_improvements = [imp for imp in improvements if imp > 0]
    print(f"\n  Epochs with loss decrease: {len(positive_improvements)}/{len(improvements)}")

    # Top-3 improvements (enumerate + sorted + unpacking)
    indexed_improvements = list(enumerate(improvements, start=2))
    top_3 = sorted(indexed_improvements, key=lambda pair: pair[1], reverse=True)[:3]
    print(f"  Top 3 improvement epochs:")
    for epoch, improvement in top_3:
        print(f"    Epoch {epoch}: loss decreased by {improvement:.4f}")

    # Convergence check (ternary + walrus)
    if (final_loss := losses[-1]) < 0.2:
        verdict = "✅ Model converged successfully!"
    elif final_loss < 1.0:
        verdict = "⚠️  Model partially converged — try more epochs"
    else:
        verdict = "❌ Model did not converge — check hyperparameters"
    print(f"\n  Verdict: {verdict}")

    # Summary table
    print(f"\n  {'Epoch':>6} {'Loss':>10} {'Accuracy':>10} {'Status':>8}")
    print(f"  {'-'*6} {'-'*10} {'-'*10} {'-'*8}")
    for epoch, (loss, acc) in enumerate(zip(losses, accuracies), start=1):
        status = "✅" if loss < 0.5 else "⏳"
        print(f"  {epoch:>6} {loss:>10.4f} {acc:>9.2%} {status:>8}")


# ===== Run it! =====
config = TrainingConfig(
    model_name="MiniTransformer-v1",
    num_epochs=15,
    batch_size=64,
    learning_rate=3e-4,
    use_gpu=True,
    patience=5,
)

history = simulate_training(config)
analyze_results(history, config)

print(f"\n🎉 Module 1 Practice Complete!")
print(f"Concepts used: variables, types, f-strings, type hints,")
print(f"conditionals, loops, enumerate, zip, comprehensions,")
print(f"unpacking, walrus operator, ternary, repr/str, print debugging")
```
